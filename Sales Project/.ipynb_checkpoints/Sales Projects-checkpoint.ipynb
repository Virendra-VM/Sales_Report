### Cleaning the Sales Data


```python
salesdb['Description'].duplicated().sum() ## Finding the Duplicate values
```

```python
cleaningsales = salesdb.drop_duplicates() ## Removing the Duplicate Values from the table
```

```python
cleaningsales['Description'].isnull().sum()
```

```python
cleaningsales = cleaningsales.dropna(subset=["Description"]) ## Removing the null rows from the table
```

```python
cleaningsales['CustomerID'].isnull().sum()
```

```python
cleaningsales = cleaningsales.dropna(subset=['CustomerID'])
```

```python
cleaningsales.isnull().sum()
```

```python
salesdb = cleaningsales 
```
